```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error
```

```python
np.random.seed(42)

x = np.linspace(0, 10, 40).reshape(-1, 1)
y = np.sin(x).ravel() + np.random.normal(0, 0.1, 40)

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 42)

simple_model = DecisionTreeRegressor(
    max_depth= 5,
    random_state= 42
)

simple_model.fit(x_train, y_train)

complex_model = DecisionTreeRegressor(
    random_state= 42
)

complex_model.fit(x_train, y_train)

simple_train = simple_model.predict(x_train)
simple_test = simple_model.predict(x_test)

complex_train = complex_model.predict(x_train)
complex_test = complex_model.predict(x_test)

print("Simple Model:")
print("Train Error:", mean_absolute_error(y_train, simple_train))
print("Test Error:", mean_absolute_error(y_test, simple_test))

print("\nComplex Model:")
print("Train Error:", mean_absolute_error(y_train, complex_train))
print("Test Error:", mean_absolute_error(y_test, complex_test))
```
